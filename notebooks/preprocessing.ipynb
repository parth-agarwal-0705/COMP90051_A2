# Preprocessing

In this notebook we will:
- preprocess and save all images from the data/raw directory into data/processed directory

```python
import os
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import cv2

src_path = os.path.abspath(os.path.join("..", "src"))
if src_path not in sys.path:
    sys.path.insert(0, src_path)
from preprocess import preprocess_image
```

```python
# read in image metadata df
parent_dir = Path.cwd().parent
df = pd.read_csv(parent_dir / "data/image_metadata.csv")
df.head()
```

```python
# processed path
processed_dir = parent_dir / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)
```

```python
for index, row in df.iterrows():
    img = preprocess_image(row["path"], 224)
    cv2.imwrite(processed_dir / row["filename"], img.astype(np.uint8))
```
